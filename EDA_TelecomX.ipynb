```python
import pandas as pd
import requests
```

```python
url ="https://raw.githubusercontent.com/ingridcristh/challenge2-data-science-LATAM/main/TelecomX_Data.json"
data = requests.get(url).json()
```

```python
df = pd.json_normalize(data)
df_final = df.copy()
```

```python
df_final.columns = df_final.columns.str.strip().str.lower().str.replace('.', '_')
```

```python
df_final['account_charges_total'] = pd.to_numeric(df_final['account_charges_total'], errors='coerce')
df_final['account_charges_monthly'] = pd.to_numeric(df_final['account_charges_monthly'], errors='coerce')

```

```python
df_final['account_contract'] = df_final['account_contract'].str.strip().str.lower()
df_final['account_paymentmethod'] = df_final['account_paymentmethod'].str.strip().str.lower()
df_final['internet_internetservice'] = df_final['internet_internetservice'].str.strip().str.lower()
```

```python
df_final['account_contract'] = df_final['account_contract'].replace({'month-to-month': 'monthly'})
```

```python
df_final['Cuentas_Diarias'] = (df_final['account_charges_monthly'] / 30).round(2)
```

```python
df_final['calculo'] = df_final['account_charges_monthly'] * df_final['customer_tenure']
df_final['diferencia'] = df_final['account_charges_total'] - df_final['calculo']
```

```python
duplicados = df_final.duplicated(subset=['customerid']).sum()
nulos = df_final.isnull().sum()
```

```python
print(f"Duplicados: {duplicados}")
print(f"Nulos por columna:\n{nulos}")
```

```python
print(df_final[['account_charges_monthly', 'Cuentas_Diarias', 'calculo', 'diferencia']].head())
```

##**Estandarización y transformación de datos**

```python
columnas_binario = ['customer_partner', 'customer_dependents']
df_final[columnas_binario] = df_final[columnas_binario].replace({'Sí':1, 'No':0})
```

```python
df_final.rename(columns={
    'customer_gender': 'genero',
    'customer_seniorcitizen': 'adulto_mayor',
    'account_contract': 'tipo_contrato',
    'account_paperlessbilling': 'sin_factura_papel',
    'account_paymentmethod': 'metodo_pago',
    'account_charges_monthly': 'cargo_mensual',
    'account_charges_total': 'cargo_total'
}, inplace=True)
```
